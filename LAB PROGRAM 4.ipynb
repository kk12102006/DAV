```python
dict={
    "name":["A","B","C","D","E","F","G","H","I"],
    "age":[10,None,12,12,None,15,16,19,20],
    "mark":[67,59,80,None,40,45,50,70,60],
    "branch":["cse",None,"ece","ece","mec","eee",None,"cse",None]
}
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(dict)
df
```

```python
age_median=df["age"].median()
print("Median:",age_median)
```

```python
mark_mean=df["mark"].mean()
print("Mean:",mark_mean)
```

```python
df["age"]=df["age"].fillna(age_median)
df
```

```python
df["mark"]=df["mark"].fillna(mark_mean)
df
```

```python
df.isnull().sum()
```

```python
df["branch"]=df["branch"].ffill()
df
```
