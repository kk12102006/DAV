```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df=pd.read_csv("iris.csv")
print(df)
```

```python
data=pd.DataFrame(df)
```

```python
print("Using SEABORN")
sns.histplot(data=df,x="petal_length",bins=30)
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Using MATPLOTLIB")
plt.hist(df["petal_length"],bins=30)
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Using SEABORN")
sns.histplot(data=df,x="sepal_length",bins=30)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Using SEABORN")
sns.histplot(data=df,x="petal_width",bins=30)
plt.title("Distribution of Petal Width")
plt.xlabel("Petal Width (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
print("Using SEABORN")
sns.histplot(data=df,x="sepal_width",bins=30)
plt.title("Distribution of Sepal Width")
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Frequency")
plt.show()
```
